```python
# Experiment 1: Data Cleaning
# Demonstrates missing-value handling, duplicate removal, and outlier detection.

import pandas as pd
import numpy as np

# Sample dataset (replace this DataFrame with pd.read_csv("your_file.csv") if needed)
df = pd.DataFrame({
    "Age": [21, 22, np.nan, 24, 25, 25, 120, 23],
    "Salary": [30000, 32000, 35000, np.nan, 40000, 40000, 50000, 31000]
})

print("Original data:")
print(df)

# 1. Handle missing numeric values using the column median
for column in df.select_dtypes(include="number").columns:
    df[column] = df[column].fillna(df[column].median())

# 2. Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# 3. Detect outliers using the IQR method and replace them with the upper/lower bounds
for column in ["Age", "Salary"]:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outlier_mask = (df[column] < lower_bound) | (df[column] > upper_bound)
    print(f"\nOutliers in {column}:")
    print(df.loc[outlier_mask, column].tolist())
    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)

print("\nCleaned data:")
print(df)
```
